# Dataset Transactions.xlsx

### Dataset transactions.xlsx_sheet transactions

```python
print("Dataset transactions.xlsx")

excel_path = 'transactions.xlsx'
if not os.path.exists(excel_path):
    raise FileNotFoundError(
        f"File '{excel_path}' tidak ditemukan. Pastikan file berada di folder."
    )

xls = pd.ExcelFile(excel_path)
print(f"Sheet dalam {excel_path}: {xls.sheet_names}")

required_sheets = {'transactions', 'customer'}
missing_sheets = required_sheets.difference(xls.sheet_names)
if missing_sheets:
    raise ValueError(f"Sheet tidak ditemukan: {sorted(missing_sheets)}")

df_trans = pd.read_excel(excel_path, sheet_name='transactions')
df_cust = pd.read_excel(excel_path, sheet_name='customer')

print("Ukuran data transaksi:", df_trans.shape)
print("Ukuran data pelanggan:", df_cust.shape)
```

```python
# Menentukan status promo sebelum menangani nilai kosong
promo_code_original = df_trans['promo_code'].copy()

df_trans['promo_status'] = np.where(
    promo_code_original.isna()
    | promo_code_original.astype('string').str.strip().eq('').fillna(True),
    'Non-Promo',
    'Promo'
)

# Menyiapkan kode promo  untuk analisis
df_trans['promo_code_clean'] = (
    promo_code_original
    .astype('string')
    .str.strip()
    .replace('', pd.NA)
    .fillna('NO_PROMO')
)

# Memastikan kolom numerik memiliki tipe data numerik
numeric_cols_trans = [
    'quantity', 'price', 'promo_amount', 'shipment_fee'
]
for col in numeric_cols_trans:
    df_trans[col] = pd.to_numeric(df_trans[col], errors='coerce')

# Nilai promo yang kosong diperlakukan sebagai tidak ada potongan promo
df_trans['promo_amount'] = df_trans['promo_amount'].fillna(0)

# Perhitungan nilai transaksi
df_trans['gross_amount'] = df_trans['quantity'] * df_trans['price']
df_trans['NMV'] = df_trans['gross_amount'] - df_trans['promo_amount']
df_trans['total_paid'] = df_trans['NMV'] + df_trans['shipment_fee']
```

```python
# Mengubah kolom tanggal menjadi datetime dan membuat kolom turunan
df_trans['created_at'] = pd.to_datetime(df_trans['created_at'], errors='coerce')

invalid_tx_dates = df_trans['created_at'].isna().sum()
print(f"Jumlah tanggal transaksi kosong/tidak valid: {invalid_tx_dates}")

df_trans['year'] = df_trans['created_at'].dt.year
df_trans['year_month'] = df_trans['created_at'].dt.to_period('M').astype('string')
df_trans['month_num'] = df_trans['created_at'].dt.month
df_trans['day_name'] = df_trans['created_at'].dt.day_name()
```

```python
df_trans.head()
```

### Dataset transaction.xlsx_sheet Customer

```python
# Membersihkan dan menyiapkan kolom tanggal pelanggan
df_cust['birthdate'] = pd.to_datetime(df_cust['birthdate'], errors='coerce')
df_cust['first_join_date'] = pd.to_datetime(df_cust['first_join_date'], errors='coerce')
df_cust['join_year'] = df_cust['first_join_date'].dt.year

df_cust['full_name'] = (
    df_cust['first_name'].fillna('').astype(str).str.strip()
    + ' '
    + df_cust['last_name'].fillna('').astype(str).str.strip()
).str.strip()
```

```python
# Menghitung perkiraan usia berdasarkan tanggal transaksi terakhir dalam dataset
ref_date = df_trans['created_at'].max()

df_cust['customer_age'] = (
    (ref_date - df_cust['birthdate']).dt.days / 365.25
).round(1)

# Usia yang tidak valid (misalnya tanggal lahir setelah tanggal referensi) tidak dikelompokkan
df_cust.loc[df_cust['customer_age'] < 0, 'customer_age'] = np.nan
df_cust['underage'] = df_cust['customer_age'] < 15

df_cust['age_segment'] = pd.cut(
    df_cust['customer_age'],
    bins=[0, 15, 25, 35, 45, np.inf],
    right=False,
    labels=[
        '<15 Thn (Underage)',
        '15-24 Thn (Gen Z)',
        '25-34 Thn (Milenial Muda)',
        '35-44 Thn (Milenial Matang)',
        '45+ Thn (Senior)'
    ]
)
```

```python
# Mencari yang aktif buyer
active_buyers = df_trans['customer_id'].unique()
df_cust['is_active_buyer'] = df_cust['customer_id'].isin(active_buyers)
```

```python
df_cust.head()
```

### Merge transactions dan customer

```python
# Menggabungkan transaksi dengan informasi pelanggan
df_merged = pd.merge(
    df_trans,
    df_cust,
    on='customer_id',
    how='left',
    validate='many_to_one'
)

df_success = df_merged[
    df_merged['payment_status'].astype('string').str.strip().str.lower().eq('success').fillna(False)
].copy()

print("Jumlah baris transaksi setelah digabung:", len(df_merged))
print("Jumlah baris transaksi berstatus Success:", len(df_success))
print("Jumlah customer_id unik pada transaksi sukses:", df_success['customer_id'].nunique())
```

# Dataset dashboard.csv

```python
# Membaca dataset retail dan membersihkan kolom teks utama
dashboard_path = 'dashboard.csv'
if not os.path.exists(dashboard_path):
    raise FileNotFoundError(
        f"File '{dashboard_path}' tidak ditemukan. Pastikan file berada di folder."
    )

df_dash = pd.read_csv(dashboard_path, sep=';')

required_dash_cols = {
    'InvoiceNo', 'StockCode', 'Description', 'InvoiceDate','Quantity', 'UnitPrice', 'Country'}
    
missing_dash_cols = required_dash_cols.difference(df_dash.columns)
if missing_dash_cols:
    raise ValueError(f"Kolom yang dibutuhkan tidak ditemukan di dashboard.csv: {sorted(missing_dash_cols)}")

# Kolom bantu dipakai untuk membersihkan nilai tanpa menggandakan kolom saat ekspor
df_dash['InvoiceNo_str'] = df_dash['InvoiceNo'].astype('string').str.strip()
df_dash['StockCode_str'] = df_dash['StockCode'].astype('string').str.strip()
df_dash['Description_clean'] = df_dash['Description'].astype('string').str.strip()

for col in ['InvoiceNo_str', 'StockCode_str', 'Description_clean']:
    df_dash[col] = df_dash[col].replace('', pd.NA)

df_dash['InvoiceNo'] = df_dash['InvoiceNo_str']
df_dash['StockCode'] = df_dash['StockCode_str']
df_dash['Description'] = df_dash['Description_clean']

# Pastikan kolom kuantitas dan harga berupa angka
df_dash['Quantity'] = pd.to_numeric(df_dash['Quantity'], errors='coerce')
df_dash['UnitPrice'] = pd.to_numeric(df_dash['UnitPrice'], errors='coerce')

print("Ukuran awal dataset retail:", df_dash.shape)
print("Jumlah InvoiceNo kosong:", df_dash['InvoiceNo'].isna().sum())
print("Jumlah Description kosong:", df_dash['Description'].isna().sum())
```

```python
df_dash.head()
```

```python
# Menandai transaksi batal dan biaya pengiriman
df_dash['is_cancelled'] = (
    df_dash['InvoiceNo_str']
    .str.startswith('C', na=False)
)
df_dash['is_postage'] = df_dash['StockCode_str'].eq('POST').fillna(False)

df_dash['tx_type'] = np.select(
    [df_dash['is_cancelled'], df_dash['is_postage']],
    ['Cancelled', 'Postage Fee'],
    default='Product Sale'
)

print("Jumlah baris transaksi batal:", int(df_dash['is_cancelled'].sum()))
print("Jumlah baris biaya pengiriman:", int(df_dash['is_postage'].sum()))
```

```python
# Mengolah tanggal invoice secara fleksibel agar jam dan variasi format tanggal tetap terbaca
df_dash['InvoiceDate_dt'] = pd.to_datetime(
    df_dash['InvoiceDate'],
    errors='coerce',
    format='mixed'
)

invalid_invoice_dates = int(df_dash['InvoiceDate_dt'].isna().sum())
print("Jumlah InvoiceDate kosong/tidak valid:", invalid_invoice_dates)

df_dash['InvoiceYear'] = df_dash['InvoiceDate_dt'].dt.year
df_dash['InvoiceMonthNum'] = df_dash['InvoiceDate_dt'].dt.month
df_dash['YearMonth'] = df_dash['InvoiceDate_dt'].dt.to_period('M').astype('string')

# Nilai penjualan kotor dihitung dari Quantity x UnitPrice.
# Nilai ini belum memperhitungkan diskon, pajak, biaya, atau lainnya.
df_dash['Gross_Revenue'] = (df_dash['Quantity'] * df_dash['UnitPrice']).round(2)

# Hanya untuk membantu analisis transaksi yang tidak dibatalkan.
# Istilah ini tidak sama dengan pendapatan bersih akuntansi.
df_dash['Active_Gross_Revenue'] = np.where(
    df_dash['is_cancelled'],
    0.0,
    df_dash['Gross_Revenue']
)
```

```python
# Menyaring baris transaksi yang tidak dibatalkan
df_dash_clean = df_dash.loc[~df_dash['is_cancelled']].copy()

print("Ukuran data awal:", df_dash.shape)
print("Ukuran data tanpa transaksi batal:", df_dash_clean.shape)
print("Jumlah invoice unik non-cancelled:", df_dash_clean['InvoiceNo_str'].nunique())
print(
    "Catatan: transaksi yang tidak dibatalkan belum tentu sama dengan pendapatan bersih, "
    "karena nilai dihitung dari Quantity x UnitPrice."
)
```

### Statistik Deskriptif

Statistik berikut merangkum transaksi berstatus `Success` pada dataset transaksi dan baris non-cancelled pada dataset retail. Keduanya merupakan dataset berbeda, sehingga nilai pendapatannya tidak dijumlahkan satu sama lain.

```python
def get_summary_stats(df, cols):
    # Menghasilkan statistik deskriptif untuk kolom numerik
    stats = df[cols].agg(['count', 'mean', 'median', 'min', 'max', 'std']).T
    return stats.round(2)

# Statistik khusus transactions.xlsx
print('Statistik — transactions.xlsx (transaksi berstatus Success)')
tx_num_cols = [
    'quantity', 'price', 'gross_amount', 'promo_amount',
    'shipment_fee', 'total_paid', 'customer_age'
]
stats_tx = get_summary_stats(df_success, tx_num_cols)
display(stats_tx)

# Statistik khusus dashboard.csv
print('Statistik — dashboard.csv (baris retail non-cancelled)')
dash_num_cols = ['Quantity', 'UnitPrice', 'Gross_Revenue']
stats_dash = get_summary_stats(df_dash_clean, dash_num_cols)
display(stats_dash)

print('Catatan: kedua tabel statistik tetap terpisah dan tidak dijumlahkan.')
```

```python
# Ekspor hasil bersih ke folder dan file terpisah per sumber data

# 1) Ekspor khusus transactions.xlsx
os.makedirs('Output Transactions', exist_ok=True)
df_merged.to_csv(
    'Output Transactions/cleaned_transactions_with_customer.csv',
    index=False, encoding='utf-8-sig'
)
df_cust.to_csv(
    'Output Transactions/cleaned_customers.csv',
    index=False, encoding='utf-8-sig'
)
stats_tx.to_csv('Output Transactions/statistik_transactions.csv', encoding='utf-8-sig')

excel_tx_output = 'Output Transactions/transactions_cleaning_and_eda.xlsx'
with pd.ExcelWriter(excel_tx_output, engine='openpyxl') as writer:
    df_trans.to_excel(writer, sheet_name='Transaksi Bersih', index=False)
    df_cust.to_excel(writer, sheet_name='Pelanggan Bersih', index=False)
    df_merged.to_excel(writer, sheet_name='Transaksi + Pelanggan', index=False)
    stats_tx.to_excel(writer, sheet_name='Statistik Transaksi', index=True)

# 2) Ekspor khusus dashboard.csv
os.makedirs('Output Retail', exist_ok=True)
helper_cols = ['InvoiceNo_str', 'StockCode_str', 'Description_clean']
df_dash_export = df_dash.drop(
    columns=[col for col in helper_cols if col in df_dash.columns],
    errors='ignore'
)
df_dash_clean_export = df_dash_clean.drop(
    columns=[col for col in helper_cols if col in df_dash_clean.columns],
    errors='ignore'
)
df_dash_clean_export.to_csv(
    'Output Retail/cleaned_retail_active.csv',
    index=False, encoding='utf-8-sig'
)
df_dash_export.to_csv(
    'Output Retail/cleaned_retail_with_flags.csv',
    index=False, encoding='utf-8-sig'
)
stats_dash.to_csv('Output Retail/statistik_retail.csv', encoding='utf-8-sig')

excel_retail_output = 'Output Retail/retail_cleaning_and_eda.xlsx'
with pd.ExcelWriter(excel_retail_output, engine='openpyxl') as writer:
    df_dash_export.to_excel(writer, sheet_name='Retail dengan Flags', index=False)
    df_dash_clean_export.to_excel(writer, sheet_name='Retail Aktif', index=False)
    stats_dash.to_excel(writer, sheet_name='Statistik Retail', index=True)

print('Ekspor selesai. Kedua sumber disimpan terpisah:')
print(f'- Dataset transactions.xlsx: {excel_tx_output}')
print(f'- Dataset dashboard.csv: {excel_retail_output}')
print('Tidak ada file ekspor gabungan yang mencampur kedua sumber data.')
```

# Visualisasi EDA 

```python
# EDA khusus transactions.xlsx: tren nilai transaksi per bulan

monthly_tx = (
    df_success.dropna(subset=['created_at'])
    .groupby('year_month', dropna=True)
    .agg(
        gross_revenue=('gross_amount', 'sum'),
        transaction_count=('booking_id', 'nunique'),
        active_customers=('customer_id', 'nunique')
    )
    .reset_index()
    .sort_values('year_month')
)

display(monthly_tx)

os.makedirs('Output Transactions/Charts', exist_ok=True)
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_tx['year_month'],
    monthly_tx['gross_revenue'],
    marker='o',
    linewidth=2
)
plt.title('Tren Gross Revenue Bulanan — transactions.xlsx')
plt.xlabel('Bulan')
plt.ylabel('Gross Revenue (satuan sesuai dataset)')
plt.xticks(rotation=60)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(
    'Output Transactions/Charts/tren_pendapatan_bulanan_transactions.png',
    dpi=200, bbox_inches='tight'
)
plt.show()
```

```python
# Mengidentifikasi bulan dengan gross revenue tertinggi dan terendah

if not monthly_tx.empty:
    highest_month = monthly_tx.loc[monthly_tx['gross_revenue'].idxmax()]
    lowest_month = monthly_tx.loc[monthly_tx['gross_revenue'].idxmin()]

    print(f"Bulan dengan gross revenue tertinggi: {highest_month['year_month']}")
    print(f"Nilai tertinggi: {highest_month['gross_revenue']:,.2f}")
    print(f"Jumlah booking unik pada bulan tertinggi: {int(highest_month['transaction_count'])}")

    print(f"\nBulan dengan gross revenue terendah: {lowest_month['year_month']}")
    print(f"Nilai terendah: {lowest_month['gross_revenue']:,.2f}")
    print(f"Jumlah booking unik pada bulan terendah: {int(lowest_month['transaction_count'])}")
else:
    print("Tidak ada data tanggal transaksi valid untuk analisis bulanan.")
```

### EDA Dataset Retail (`dashboard.csv`)

Bagian ini menganalisis dataset retail secara terpisah. Analisis pendapatan menggunakan `Quantity × UnitPrice` pada baris yang tidak dibatalkan dan bukan biaya pengiriman. Nilai ini disebut **gross sales value**, bukan pendapatan bersih akuntansi. Baris dengan kuantitas negatif, jika ada, tetap memengaruhi total sebagai penyesuaian/retur.

```python
# Menyiapkan baris penjualan produk untuk EDA retail
df_retail_sales = df_dash_clean.loc[
    (~df_dash_clean['is_postage'])
    & df_dash_clean['InvoiceDate_dt'].notna()
    & df_dash_clean['Quantity'].notna()
    & df_dash_clean['UnitPrice'].notna()
].copy()

print('Jumlah baris retail aktif untuk analisis:', len(df_retail_sales))
print('Invoice unik:', df_retail_sales['InvoiceNo_str'].nunique())
print('Negara unik:', df_retail_sales['Country'].nunique())
print('Deskripsi produk kosong:', df_retail_sales['Description'].isna().sum())
print('Rentang tanggal:', df_retail_sales['InvoiceDate_dt'].min(), 'sampai', df_retail_sales['InvoiceDate_dt'].max())
```

```python
# EDA khusus dashboard.csv: tren gross sales value per bulan
monthly_retail = (
    df_retail_sales.groupby('YearMonth', dropna=True)
    .agg(
        gross_sales_value=('Gross_Revenue', 'sum'),
        invoice_count=('InvoiceNo_str', 'nunique'),
        units_sold=('Quantity', 'sum')
    )
    .reset_index()
    .sort_values('YearMonth')
)
display(monthly_retail)

os.makedirs('Output Retail/Charts', exist_ok=True)
plt.figure(figsize=(12, 5))
plt.plot(monthly_retail['YearMonth'], monthly_retail['gross_sales_value'], marker='o', linewidth=2)
plt.title('Tren Gross Sales Value Bulanan — dashboard.csv')
plt.xlabel('Bulan')
plt.ylabel('Gross Sales Value (sesuai satuan dataset)')
plt.xticks(rotation=60)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('Output Retail/Charts/tren_gross_sales_retail.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# EDA khusus dashboard.csv: 10 produk dengan gross sales value tertinggi
top_products = (
    df_retail_sales.dropna(subset=['Description'])
    .groupby('Description', as_index=False)
    .agg(gross_sales_value=('Gross_Revenue', 'sum'), units_sold=('Quantity', 'sum'))
    .sort_values('gross_sales_value', ascending=False)
    .head(10)
)
display(top_products)

plt.figure(figsize=(11, 6))
sns.barplot(data=top_products, y='Description', x='gross_sales_value')
plt.title('Top 10 Produk Berdasarkan Gross Sales Value — dashboard.csv')
plt.xlabel('Gross Sales Value')
plt.ylabel('Produk')
plt.tight_layout()
plt.savefig('Output Retail/Charts/top_10_produk_retail.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# EDA khusus dashboard.csv: 10 negara dengan gross sales value tertinggi
top_countries = (
    df_retail_sales.dropna(subset=['Country'])
    .groupby('Country', as_index=False)
    .agg(gross_sales_value=('Gross_Revenue', 'sum'), invoice_count=('InvoiceNo_str', 'nunique'))
    .sort_values('gross_sales_value', ascending=False)
    .head(10)
)
display(top_countries)

plt.figure(figsize=(10, 6))
sns.barplot(data=top_countries, y='Country', x='gross_sales_value')
plt.title('Top 10 Negara Berdasarkan Gross Sales Value — dashboard.csv')
plt.xlabel('Gross Sales Value')
plt.ylabel('Negara')
plt.tight_layout()
plt.savefig('Output Retail/Charts/top_10_negara_retail.png', dpi=200, bbox_inches='tight')
plt.show()
```
